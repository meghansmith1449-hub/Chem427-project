**Question:** Does AFM height read silica nanoparticles smaller than SEM and SAXS do, while reading gold nanoparticles correctly, by more than the combined uncertainty?

**Data:** nPSize image and annotation records, Zenodo (2021–2022): https://doi.org/10.5281/zenodo.5578927, https://doi.org/10.5281/zenodo.5578878, https://doi.org/10.5281/zenodo.5577401, https://doi.org/10.5281/zenodo.5886834, https://doi.org/10.5281/zenodo.7016466. Accessed November 8, 2026.

**Project:** 11. Nanoparticle size by three techniques

```python
import json
import zipfile
import pandas as pd

with zipfile.ZipFile("data/annotations.zip") as z:
    annotations = json.load(z.open("annotations.json"))

rows = []
for lab in annotations.values():
    for mkey, scope in lab.items():
        if not mkey.startswith("microscope"):
            continue
        for image in scope["data"].values():
            for pkey, particle in image["particles"].items():
                if not pkey.startswith("particle"):
                    continue
                row = {
                    "laboratory": lab["name"],
                    "technique": scope["type"],
                    "npsize": image["npsize"],
                    "image": image["name"],
                    "pixel_size": image["pixel_size"],
                }
                row.update(pd.json_normalize(particle["measurements"]).iloc[0].to_dict())
                rows.append(row)

data = pd.DataFrame(rows)
print(f"{len(data)} rows")
print(list(data.columns))
print("particle size (Feret diameter, AFM height): nm")
```
